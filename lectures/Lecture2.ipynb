```python
import matplotlib.pyplot as plt
```

```python
y = 3  # координата
yd = 1  #  скорость
m = 5  # масса
dt = 0.01  # период дискретизации симуляции
```

```python
iter_num = 1000  # число шагов симуляции
yhist, ydhist, uhist, thist = [], [], [], []  # логи(история всего)
```

```python
Kp, Kd = 30, 50

for i in range(iter_num):
    u = - Kp * y - Kd * yd

    ydd = u / m

    y += yd * dt
    yd += ydd * dt

    yhist.append(y)
    ydhist.append(yd)
    uhist.append(u)
    thist.append(i * dt)
```

```python
plt.plot(thist, yhist)
plt.show()
plt.plot(thist, ydhist)
plt.show()
plt.plot(thist, uhist)
plt.show()
plt.plot(yhist, ydhist)
plt.show()
```
